```python
# ── Cell 2: Clone your GitHub repo ────────────────────────────────────────
import os

GITHUB_USERNAME = 'Zahira1'
REPO_NAME       = 'DeepLearning'
BRANCH          = 'homework3-branch'

repo_url = f'https://github.com/{GITHUB_USERNAME}/{REPO_NAME}.git'

if not os.path.exists(f'/content/{REPO_NAME}'):
    os.system(f'git clone -b {BRANCH} {repo_url} /content/{REPO_NAME}')
    print('Cloned successfully!')
else:
    print('Repo already cloned — run Cell 5 to pull latest changes')

# Files are at the root of the branch (no subfolder)
os.chdir(f'/content/{REPO_NAME}')
print('Working directory:', os.getcwd())
print(os.listdir('.'))
```

```python
# ── Cell 3: Unzip data from Drive ─────────────────────────────────────────
import zipfile, os

os.chdir('/content/DeepLearning')

# Adjust path if you put the zip somewhere else in Drive
DATA_ZIP = '/content/drive/MyDrive/hw3_data/classification_data.zip'

if not os.path.exists('classification_data'):
    print('Unzipping data...')
    with zipfile.ZipFile(DATA_ZIP, 'r') as z:
        z.extractall('.')
    print('Done!')
else:
    print('Data already extracted')

print(os.listdir('.'))
```
